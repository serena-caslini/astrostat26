```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
import urllib.request
urllib.request.urlretrieve("https://raw.githubusercontent.com/nshaud/ml_for_astro/main/stars.csv", "stars.csv")
```

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
import seaborn as sns
```

```python
df = pd.read_csv("stars.csv")
```

```python
df
```

```python
le = LabelEncoder()
# Assign unique integers from 0 to 6 to each star type
df['Star type'] = le.fit_transform(df['Star type'])
labels = le.inverse_transform(df['Star type'])
class_names = le.classes_
print(class_names)
```

```python
fig = plt.figure(figsize=(7, 7))
sns.scatterplot(data=df, x='Temperature (K)', y='Luminosity(L/Lo)', hue=labels, palette='tab10')

plt.xscale('log')
plt.yscale('log')
plt.xticks([5000, 10000, 50000])
plt.xlim(5e4, 1.5e3)
plt.show()
```

```python
data = np.array(df.iloc[:,:4])
means = np.mean(data, axis=0)
```

```python
print(means)
```

```python
scaler = StandardScaler()
scaler.fit(data)
data = scaler.transform(data)
```

```python
pca = PCA(n_components=2) # n_components can be optionally set
pca.fit(data) 
```

```python
eigenvectors = pca.components_ 
eigenvalues = pca.explained_variance_ratio_ 
```

```python
eigenvalues
```

```python
data_pca = pca.transform(data)
```

```python
np.shape(data_pca)
```

```python
plt.scatter(data_pca[:,0], data_pca[:,1], ls = '', marker = 'o')
```

```python
data_reconstruct = scaler.inverse_transform(pca.inverse_transform(data_pca))
```

```python
fig = plt.figure(figsize=(7, 7))
plt.loglog(data_reconstruct[:,0], data_reconstruct[:,1], ls = '', marker = 'o')
plt.xlim(5e4, 1.5e3)
```
